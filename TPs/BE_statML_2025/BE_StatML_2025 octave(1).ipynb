<span style="color:lightblue">
Afin de savoir si l'effet est différent en fonction de la marque du moteur, nous allons réaliser un teste de fisher sur le modèle de groupe de la regression linéaire. L'hypothèse H0 est que les pente de chaque groupe sont égale. Pour ce faire, on défini le modèle de la regression par groupe en prenant comme groupe de référence la brand1. Ensuite on définie le modèle de regression par groupe en faisant en sorte que le betha_1 prime de la brand2 sois a 0. Pour ce faire, on vas donner à la regression sur le groupe 2 la même  pente sur sur le groupe 1. Nous faisont ensuite les calcule de y.  
Pour effectuer notre teste de fisher, nous faisont le calcule des variance des erreurs de chacun de ces modèle (considérer comme normale). Nous faisont le rapport de ces deux estimateur de variance avant d'utiliser la fonction de répartition de la loi de fisher (donner par stats.f.cdf) afin de calculer la p-value qui nous fera accepter ou non l'hyptèse H0.<span>  

```python
from scipy.stats import f

# On calcule les résidut du modèle complet grâce au calcule de régression précedent
residu_complet = np.sum((y_pred_brand1 - y_brand1) ** 2) + np.sum((y_pred_brand2 - y_brand2) ** 2)

# On cherche à calculer le modèle réduit avec les pentes des deux groupe égale mais pas les résidut
# Pour ce faire on choisi le groupe 1 comme groupe de référence
# Afin de données la même pente mais pas la même ordonner a l'origine, on vas ajouter une colonne au dataframe
# remplie de 0 pour la brand1 et de 1 pour la brand2. Ainsi quand le modèle fera la regression linéaire,
# il calculera la pente sur tout les x et y mais l'ordonné a l'origine sera calculer sur la brand2, ce qui nous donnera des 
# ordonnée à l'origine différente pour les deux brand


df = pd.concat([df_brand1.assign(Marque=0), df_brand2.assign(Marque=1)], axis=0) # Ajout de la colonne de  et de 

x_reduit = np.column_stack([df["Concentration"], df["Marque"]]) # Récupération du x

# Entrainement du modèle
model_reduit = LinearRegression()

model_reduit.fit(x_reduit, np.array(df["Efficiency"]))
y_pred_reduit = model_reduit.predict(x_reduit)

# Calcule du résidut
residu_reduit = np.sum((y_pred_reduit - np.array(df["Efficiency"])) ** 2)

# Ensuite on fait un teste de Fisher sur les résidut pour savoir si les pentes sont égale ou non
# On utilise la formule : F = (SCE / ddl_SCE) / (SRC / ddl_SCR)

n = len(df) # nombre de paramètres estimés
ddl_SCE = 2 - 1 # On fait K - 1 avec K le nombre de groupe, cela donne 2 - 1
ddl_SCR = n - 2 # On a N - K avec N le nombre totale d'observation

SCE = (residu_reduit - residu_complet)
SRC = residu_complet

F_stat = (SCE / ddl_SCE) / (SRC / ddl_SCR)

p_value = 1 - f.cdf(F_stat, ddl_SCE, ddl_SCR) # Calcule de la p-value



print(f"F_stat = {F_stat}")
print(f"p-value = {p_value:.6f}")
if p_value < 0.05:
    print("Comme la p-value est en dessous de notre seuil alpha (5%), on rejette H0, les pentes ne sont pas égale.")
else:
    print("Comme la p-value est au dessus de notre seuil alpha (5%), on accepte H0, les pentes sont égale..")
```

```python
from sklearn.model_selection import KFold

from sklearn.metrics import mean_squared_error

def K_fold(X,y):
    kf = KFold(n_splits=8, shuffle=True, random_state=42)

    pente_list = []
    origine_list = []
    mse_score = []
    # On sépare le jeu de données en train et en test
    for train_index, test_index in kf.split(X):
        X_train, X_test = X[train_index], X[test_index]
        y_train, y_test = y[train_index], y[test_index]    
        
        
        lr = LinearRegression()
        lr.fit(X_train, y_train)

        # On garde en mémoire les données de la rgression linéaire ainsi que la MSE
        pente_list.append(lr.coef_)
        origine_list.append(lr.intercept_)

        y_pred = lr.predict(X_test)
        mse_score.append(mean_squared_error(y_test, y_pred))
    

    mse = np.mean(mse_score)
    # on prend le meilleur modèles
    if mse >= np.min(mse_score):

        indice = np.argmin(mse_score)
        lr.coef_, lr.intercept_ = np.array(pente_list[indice]), np.array(origine_list[indice])
        mse = np.min(mse_score)
    else:
        pente_fin = np.mean(pente_list)
        origine_fin = np.mean(origine_list)

        lr.coef_, lr.intercept_ = np.array([pente_fin]), np.array([origine_fin])    

    return lr, mse
```

```python
from sklearn.metrics import r2_score, mean_squared_error

selected_sensors = [1,12,15]

for sensor in selected_sensors :

    x_sensor = np.array(X.iloc[:,sensor-1]).reshape(-1,1)
    y = np.array(Y).reshape(-1)


    lr_sensor, mean_mse = K_fold(x_sensor,y)

    y_pred_sensor = lr_sensor.predict(x_sensor)

    print("R² :", r2_score(Y, y_pred_sensor))

    # Autres métriques
    print("MSE :", mean_squared_error(Y, y_pred_sensor))

    plt.figure()
    plt.scatter(x_sensor,Y,color='blue')
    plt.plot(x_sensor,y_pred_sensor,color='red', label=f"y = {lr_sensor.coef_[0]:.3f} x + {lr_sensor.intercept_:.3f}")
    plt.title(f"Régression linéaire de la fatigue des ailes de l'avion en fonction des valeurs du capteur n°{sensor}")
    plt.xlabel(f"Valeurs du capteur n°{sensor}")
    plt.ylabel(f"Fatigue des ailes de l'avion")
    plt.legend()
    plt.show()
```

```python
def cook(x, y, i):
    lr_glob = LinearRegression()
    
    lr_glob.fit(x.reshape(-1,1),y)

    lr_i = LinearRegression()

    x_del_i = np.delete(x,i).reshape(-1,1)
    y_del_i = np.delete(y,i).reshape(-1,1)

    lr_i.fit(x_del_i, y_del_i)

    y_hat = lr_glob.predict(x.reshape(-1,1)).reshape(-1)
    y_hat_i = lr_i.predict(x.reshape(-1,1)).reshape(-1)

    s2 = (1/(len(y) - 1)) * np.sum((y - y_hat)**2)
    numerateur = np.sum((y_hat_i - y_hat)**2)
    
    return numerateur / (s2 * 2)


distance_cook = []
for i in range(len(y)):
    distance_cook.append(cook(np.array(X['sensor_12']),y,i))
    print(f"la distance de cook pour i = {i} est {distance_cook[i]}")

plt.plot(np.linspace(1,len(y),len(y)), distance_cook)
plt.show()
```

<span style="color:lightblue">
On remarque bien la présence des 2 outliers dans les données du capteur n°12, ayant une valeur de -10.82 et -9.21 respectivement, alors que la moyenne des valeurs du capteur est de -3.01.  
Si l'on suppose que les erreurs de prédiction suivent une loi normale centrée, on peut également procéder à un test de Student pour détecter les outliers.
</span>

<span style="color:lightblue">
L'hypothèse H0 est ici que le résidu est sensiblement proche de 0, et on va utiliser la statistique du test de Student pour l'accepter ou la rejeter. Si la statistique de test est supérieure au seuil de confiance à 95% (en valeur absolue), on rejette H0, et on peut alors estimer que le point considéré est un outlier.
</span>

```python
model = sm.OLS(Y, X['sensor_12']).fit()

residuals = model.resid
sigma = residuals.std(ddof=X.shape[1])

# seuil pour test bilatéral à 5%, puisque pour une normale standard, 95% des valeurs sont dans [-1.96, 1.96].
threshold = 1.96
t = np.abs(residuals / sigma)
outliers = np.where(t > threshold)[0]

print("Indices des outliers détectés :", outliers)
```

<span style='color:lightblue'>
Question 2.4.2 : On va donc effectuer une régression linéaire régularisée. On choisit d'utiliser la méthode Lasso. Pour ce faire, il nous faut d'abord trouver le alpha optimal à utiliser dans la méthode. On va donc procéder à une cross-validation KFold, sur différentes valeurs de alpha, afin de trouver celle qui minimise la MSE. Une fois que ce sera fait, on pourra alors créer notre modèle Lasso optimal (avec la bonne valeur de alpha), et regarder quelles variables explicatives ont un coefficient non nul dans la régression. On calculera également la MSE ou le R² pour évaluer la pertinence du modèle obtenu.
</span>

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import KFold

X = pd.read_csv('./E2_sensor_vals.csv', sep=';', decimal = '.')

Y = pd.read_csv('./E2_stress_vals.csv', sep=';', decimal = '.')

# On enlève les 2 valeurs aberrantes du jeu de données
indices_to_drop = outliers

X.drop(indices_to_drop, inplace=True)
Y.drop(indices_to_drop, inplace=True)

# On centre et réduit le jeu de données
X = (X - X.mean()) / X.std()

Y = (Y - Y.mean()) / Y.std()

# On procède à la méthode de validation croisée KFold pour trouver le alpha optimal
kf = KFold(n_splits=8, shuffle=True, random_state=42)
list_alpha = np.linspace(0.001,0.99,100)
mean_mse_par_alpha = []

for alpha in list_alpha :
    # Pour chaque alpha considéré, on va construire un modèle Lasso, en distinguant
    # un ensemble d'entraînement (ici 7/8e du jeu de données) et un ensemble de test (le dernier 1/8e)
    mse_scores = []
    for train_idx, test_idx in kf.split(X):
        X_train, X_test = X.iloc[train_idx,:], X.iloc[test_idx,:]
        y_train, y_test = Y.iloc[train_idx], Y.iloc[test_idx]
        
        # Pour chaque combinaison entraînement/test, on établit le modèle Lasso avec le alpha considéré
        model = Lasso(alpha=alpha, max_iter=10000)
        model.fit(X_train, y_train)
        
        # On calcule la MSE du modèle
        y_pred = model.predict(X_test)
        mse = mean_squared_error(y_test, y_pred)
        mse_scores.append(mse)
    
    # Pour chaque alpha, on fait la moyenne des MSE obtenus lors de la réalisation de la méthode KFold
    mean_mse = np.mean(mse_scores)
    mean_mse_par_alpha.append(mean_mse)

# Sélection du meilleur alpha, i.e. celui qui minimise la MSE
best_alpha = list_alpha[np.argmin(mean_mse_par_alpha)]
print("\nMeilleur alpha =", best_alpha)

# On construit finalement le modèle Lasso optimal avec le alpha choisi
best_model = Lasso(alpha=best_alpha, max_iter=10000)
best_model.fit(X,Y)

# On estime alors les valeurs de Y à partir du modèle, et on calcule la MSE et le R² pour 
# estimer la pertinence du modèle
Y_pred = best_model.predict(X)
mse_score_lasso = mean_squared_error(Y, Y_pred)
r2 = best_model.score(X, Y)
print("R² :", r2)

print("MSE :", mse_score_lasso)
print("Coefficients du modèle Lasso optimal :",best_model.coef_)
```

<span style='color:lightblue'>
On remarque ainsi que les 3 seuls capteurs pour lesquels le coefficient dans la régression est non nul, sont les capteurs 1, 12 et 15. Cela corrobore ce que l'on avait remarqué dans les questions précédentes, à savoir que ce sont les seuls qui semblent avoir une corrélation avec le niveau de fatigue dans les ailes de l'avion. Le R² est de 0.8, ce qui est plutôt satisfaisant, et la MSE de 0.19. On peut en conclure que l'utilisation des données des capteurs, en enlevant les outliers, paraît fiable pour estimer le niveau de fatigue des ailes de l'avion.
